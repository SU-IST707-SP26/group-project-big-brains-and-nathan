```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('../data/merged.csv')
data
```

```python
data.describe()
```

```python
nfl = data[data['pick'].notnull()]
nfl
```

```python
positions = data['position_x'].value_counts().sort_values()
nfl_positions = nfl['position_x'].value_counts().sort_values()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

positions.plot(kind = 'barh', ax = ax1)
ax1.set_title('NCAA Distribution of Positions')
ax1.set_xlabel('Number of Players')
ax1.set_ylabel('Position')

nfl_positions.plot(kind = 'barh', ax = ax2)
ax2.set_title('NFL Drafted Distribution of Positions')
ax2.set_xlabel('Number of Players')
ax2.set_ylabel('Position')
plt.show()
```

```python
positions = data['conference'].value_counts().sort_values()
nfl_positions = nfl['conference'].value_counts().sort_values()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

positions.plot(kind = 'barh', ax = ax1)
ax1.set_title('Total NCAA Conference Distribution')
ax1.set_xlabel('Number of Players')
ax1.set_ylabel('Position')

nfl_positions.plot(kind = 'barh', ax = ax2)
ax2.set_title('NFL Drafted Conference Distribution')
ax2.set_xlabel('Number of Players')
ax2.set_ylabel('Position')
plt.show()
```

```python
#top 15 teams by player count
positions = data['team_x'].value_counts().head(15).sort_values()
nfl_positions = nfl['team_x'].value_counts().head(15).sort_values()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

positions.plot(kind = 'barh', ax = ax1)
ax1.set_title('Total NCAA Top 15 Teams by Amount of Players')
ax1.set_xlabel('Number of Players')
ax1.set_ylabel('Conference')

nfl_positions.plot(kind = 'barh', ax = ax2)
ax2.set_title('NFL Drafted Top 15 Teams by Amount of Players')
ax2.set_xlabel('Number of Players')
ax2.set_ylabel('Conference')
plt.show()
```

```python
#season distribution
plt.hist(data['height'])
plt.title('Age Distribution of Drafted Players')
plt.xlabel('Height (Inches)')
plt.ylabel('Number of Players')
plt.show()
```

```python
#age distribution - lot of NAs due to being NFL data
plt.hist(data['age'])
plt.title('Age Distribution of Drafted Players')
plt.xlabel('Age')
plt.ylabel('Number of Players')
plt.show()
```

```python
#usage rate distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.hist(data['usage_overall'])
ax1.set_title('Usage Distribution of All NCAA Players')
ax1.set_xlabel('Usage Rate')
ax1.set_ylabel('Number of Players')

ax2.hist(nfl['usage_overall'])
ax2.set_title('Usage Distribution of Drafted Players')
ax2.set_xlabel('Usage Rate')
ax2.set_ylabel('Number of Players')
plt.show()
```

```python
#total ppa distribution 
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.hist(data['total_ppa_all'])
ax1.set_title('PPA Distribution of All NCAA Players')
ax1.set_xlabel('Total PPA')
ax1.set_ylabel('Number of Players')

ax2.hist(nfl['total_ppa_all'])
ax2.set_title('PPA Distribution of Drafted Players')
ax2.set_xlabel('Total PPA')
ax2.set_ylabel('Number of Players')
plt.show()
```

```python
#Distribution of Passing TDs by QB
qbs = data[data['position_x'] == 'QB']
nfl_qbs = nfl[nfl['position_x'] == 'QB']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.hist(qbs['passing_TD'])
ax1.set_title('Passing TDs by All NCAA Players')
ax1.set_xlabel('Total Passing TDs')
ax1.set_ylabel('Number of Players')

ax2.hist(nfl_qbs['passing_TD'])
ax2.set_title('Passing TDs Drafted Players')
ax2.set_xlabel('Total Passing TDs')
ax2.set_ylabel('Number of Players')
plt.show()
```

```python
#QBs passing TD vs rushing TD
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.scatter(qbs['passing_TD'], qbs['rushing_TD'])
ax1.set_title('All NCAA Quarterbacks: Passing TDs vs. Rushing TDs')
ax1.set_xlabel('Total Passing TDs')
ax1.set_ylabel('Total Rushing TDs')

ax2.scatter(nfl_qbs['passing_TD'], nfl_qbs['rushing_TD'])
ax2.set_title('Drafted Quarterbacks: Passing TDs vs. Rushing TDs')
ax2.set_xlabel('Total Passing TDs')
ax2.set_ylabel('Total Rushing TDs')
plt.show()
```

```python
rbs = data[data['position_x'] == 'RB']
nfl_rbs = nfl[nfl['position_x'] == 'RB']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.scatter(rbs['rushing_TD'], rbs['receiving_TD'])
ax1.set_title('All NCAA Runningbacks: Rushing TDs vs. Receiving TDs')
ax1.set_xlabel('Total Rushing TDs')
ax1.set_ylabel('Total Receiving TDs')

ax2.scatter(nfl_rbs['rushing_TD'], nfl_rbs['receiving_TD'])
ax2.set_title('Drafted Runningbacks: Rushing TDs vs. Receiving TDs')
ax2.set_xlabel('Total Rushing TDs')
ax2.set_ylabel('Total Receiving TDs')
plt.show()
```
